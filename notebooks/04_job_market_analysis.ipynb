```python
salary = jobs["med_salary"].replace(0, np.nan)
salary_clean = salary.dropna()

salary_clean = salary_clean[
    (salary_clean >= 10000) &
    (salary_clean <= 300000)
]
jobs["med_salary"].describe()
```

```python
plt.figure(figsize=(10,5))

plt.hist(
    salary_clean,
    bins=50
)

plt.title("Salary Distribution")

plt.xlabel("Median Salary")

plt.ylabel("Number of Jobs")

plt.show()
```

```python
salary_analysis = pd.DataFrame({
    "median_salary": salary_clean
})

salary_analysis.to_csv(
    "../data/salary_analysis.csv",
    index=False
)
```

```python
salary_by_role = (
    jobs.groupby("title")["med_salary"]
    .mean()
    .sort_values(ascending=False)
    .head(20)
)

salary_by_role
```

```python
plt.figure(figsize=(12,6))

salary_by_role.plot(kind="bar")

plt.title("Top 20 Highest Paying Job Roles")

plt.xlabel("Job Title")

plt.ylabel("Average Salary")

plt.xticks(rotation=90)

plt.show()
```

```python
salary_by_role = (
    jobs.groupby("title")
    .agg(
        avg_salary=("med_salary","mean"),
        count=("title","count")
    )
)

salary_by_role = salary_by_role[
    salary_by_role["count"] >= 20
]

salary_by_role = salary_by_role.sort_values(
    "avg_salary",
    ascending=False
)

salary_by_role.head(15)
```

```python
plt.figure(figsize=(12,6))

salary_by_role.head(15).sort_values(
    "avg_salary"
)["avg_salary"].plot(
    kind="barh"
)

plt.title("Highest Paying Job Roles")
plt.xlabel("Average Salary")

plt.show()
```

```python
jobs["is_remote"] = jobs["remote_allowed"].fillna(0)

jobs["is_remote"] = jobs["is_remote"].apply(
    lambda x: "Remote" if x == 1 else "Non-Remote"
)
```

```python
remote_salary = (
    jobs.groupby("is_remote")["med_salary"]
    .mean()
)

remote_salary
```

```python
plt.figure(figsize=(6,4))

remote_salary.plot(
    kind="bar"
)

plt.title("Average Salary: Remote vs Non-Remote Jobs")
plt.ylabel("Average Salary")

plt.show()
```
