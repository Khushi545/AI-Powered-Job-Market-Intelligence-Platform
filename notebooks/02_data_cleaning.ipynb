```python
import pandas as pd
import numpy as np
import re

import warnings
warnings.filterwarnings("ignore")
```

```python
jobs = pd.read_csv("../data/job_postings.csv")

```

```python
jobs.head()
```

```python
jobs.shape
```

```python
jobs.columns
```

```python
columns_to_keep = [
    "job_id",
    "company_name",
    "title",
    "description",
    "location",
    "formatted_work_type",
    "formatted_experience_level",
    "min_salary",
    "max_salary",
    "med_salary",
    "normalized_salary",
    "pay_period",
    "currency",
    "remote_allowed",
    "skills_desc"
]
jobs = jobs[columns_to_keep]
```

```python
jobs.shape
```

```python
jobs.isnull().sum()
```

```python
text_columns = [
    "company_name",
    "description",
    "location",
    "formatted_work_type",
    "formatted_experience_level",
    "skills_desc"
]

for col in text_columns:
    jobs[col] = jobs[col].fillna("Not Available")
```

```python
salary_columns = [
    "min_salary",
    "max_salary",
    "med_salary"
]

jobs[salary_columns] = jobs[salary_columns].fillna(0)
```

```python
jobs.duplicated().sum()
```

```python
jobs = jobs.drop_duplicates()
```

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r"<.*?>", " ", text)
    text = re.sub(r"[^a-zA-Z0-9\s]", " ", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()
```

```python
jobs["clean_description"] = jobs["description"].apply(clean_text)
```

```python
jobs["clean_title"] = jobs["title"].apply(clean_text)
```

```python
jobs.head()
```

```python
jobs.info()
```

```python
jobs.isnull().sum()
```

```python
jobs["currency"] = jobs["currency"].fillna("Not Available")

jobs["remote_allowed"] = jobs["remote_allowed"].fillna(0)
```

```python
jobs.isnull().sum()
```

```python
jobs.to_csv("../data/jobs_cleaned.csv", index=False)
```

```python
import pandas as pd

# Load raw file (has normalized_salary, pay_period)
raw = pd.read_csv("../data/job_postings.csv", usecols=["job_id", "normalized_salary", "pay_period"])

# Load your current skills file
jobs = pd.read_csv("../data/jobs_with_skills_v2.csv")

# If it's already been merged before, drop old copies first to avoid duplicate columns
jobs = jobs.drop(columns=["normalized_salary", "pay_period"], errors="ignore")

# Merge in the two columns
jobs = jobs.merge(raw, on="job_id", how="left")

# Confirm they're actually there before reordering
print("Columns after merge:", jobs.columns.tolist())

# Reorder to match your Postgres table
column_order = [
    "job_id", "company_name", "title", "description", "location",
    "formatted_work_type", "formatted_experience_level",
    "min_salary", "max_salary", "med_salary", "currency", "remote_allowed",
    "skills_desc", "clean_description", "clean_title", "extracted_skills",
    "skill_count", "normalized_salary", "pay_period"
]
jobs = jobs[column_order]

jobs.to_csv("../data/jobs_with_skills_v2.csv", index=False)
print("Saved. Final columns:", jobs.columns.tolist())
```
