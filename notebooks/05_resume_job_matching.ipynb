```python
with open("../data/sample_resume.txt", "r", encoding="utf-8") as f:
    resume_text = f.read()

print(resume_text)
```

```python
skills = [
    "python",
    "sql",
    "machine learning",
    "data analysis",
    "excel",
    "power bi",
    "tableau",
    "aws",
    "docker",
    "tensorflow",
    "pandas",
    "numpy"
]
```

```python
def extract_skills(text, skills_list):

    text = text.lower()

    found = []

    for skill in skills_list:
        if skill.lower() in text:
            found.append(skill)

    return list(set(found))
```

```python
resume_skills = extract_skills(
    resume_text,
    skills
)

print(resume_skills)
```

```python
row = 26
import ast

job_skills = ast.literal_eval(
    jobs.iloc[row]["extracted_skills"]
)

print(job_skills)
```

```python
print("Resume Skills:")
print(resume_skills)

print("\nJob Skills:")
print(job_skills)
```

```python
matched_skills = set(resume_skills).intersection(job_skills)

if len(job_skills) > 0:
    match_score = (len(matched_skills) / len(job_skills)) * 100
else:
    match_score = 0

print(f"Match Score: {match_score:.2f}%")
```

```python
missing_skills = set(job_skills) - set(resume_skills)

print("Matched Skills:")
print(matched_skills)

print("\nMissing Skills:")
print(missing_skills)
```

```python
if match_score >= 80:
    recommendation = "Strong Match"
elif match_score >= 60:
    recommendation = "Moderate Match"
else:
    recommendation = "Weak Match"

print("\nRecommendation:", recommendation)
```
